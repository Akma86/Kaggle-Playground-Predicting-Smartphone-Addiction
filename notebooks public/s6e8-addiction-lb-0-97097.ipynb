```python
# ==========================================
# 🛠️ SIMPLE BLEND CODE
# ==========================================

YOUR_SUBMISSION_PATH = None  # e.g., "../input/your-notebook/submission.csv"
YOUR_BLEND_WEIGHT = 0.5      # Give your model 50% weight

if YOUR_SUBMISSION_PATH is not None:
    user_sub = pd.read_csv(YOUR_SUBMISSION_PATH)
    
    final_blend = submission.copy()
    final_blend["addicted_label"] = (submission["addicted_label"] * (1 - YOUR_BLEND_WEIGHT)) + (user_sub["addicted_label"] * YOUR_BLEND_WEIGHT)
    
    final_blend.to_csv("submission.csv", index=False)
    print("🚀 Successfully generated your blended submission.csv!")
else:
    # Default to outputting just your high score if they run it standalone
    # submission.to_csv("submission.csv", index=False)
    print("ℹ️ Outputting the default submission file.")
```
